```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
performers = pd.read_csv("performer_demographic_meta.csv")
```

```python
events = pd.read_csv("historical_events.csv")
grouped_events = events.groupby("month_code").mean(numeric_only=True).reset_index()
headings = grouped_events["month_code"].tolist()
headings = [month.replace("Month", "").replace(" ", "") for month in headings]


plt.bar(headings,grouped_events["ticket_revenue"], color='skyblue', width= 0.5)

# 3. Add labels and a title
plt.title('Revenue per Month')
plt.xlabel('Month',labelpad = 2)
plt.ylabel('Revenue (Units)')

# 4. Display the chart
plt.show()
```

```python
performer_events = events.groupby("performer_id").mean(numeric_only=True).reset_index()
#display(performer_events)
headings_performer = performer_events["performer_id"].tolist()
headings_performer = [perfomer.replace("P", "").replace(" ", "") for perfomer in headings_performer]
#print(headings_performer)

plt.bar(headings_performer,performer_events["ticket_revenue"], color='skyblue', width=0.5)

# 3. Add labels and a title
plt.title('Revenue per performer')
plt.xlabel('Perfomer')
plt.ylabel('Revenue (Units)')

# 4. Display the chart
plt.show()
```

```python
type_events = events.groupby("event_type").mean(numeric_only=True).reset_index()
#display(performer_events)
headings_type = type_events["event_type"].tolist()
#print(headings_performer)

plt.bar(headings_type,type_events["ticket_revenue"], color='skyblue', width=0.5)

# 3. Add labels and a title
plt.title('Revenue per event type')
plt.xlabel('Type')
plt.ylabel('Revenue (Units)')

# 4. Display the chart
plt.show()
```

```python
genre_events = events.groupby("genre").mean(numeric_only=True).reset_index()
#display(performer_events)
headings_genre = genre_events["genre"].tolist()
#print(headings_performer)

plt.bar(headings_genre,genre_events["ticket_revenue"], color='skyblue', width=0.5)

# 3. Add labels and a title
plt.title('Revenue per genre')
plt.xlabel('Genre')
plt.ylabel('Revenue (Units)')

# 4. Display the chart
plt.show()
```

```python
time_events = events.groupby("time_of_day").mean(numeric_only=True).reset_index()
headings_time = time_events["time_of_day"].tolist()
#print(headings_performer)

plt.bar(headings_time,time_events["ticket_revenue"], color='skyblue', width=0.5)

# 3. Add labels and a title
plt.title('Revenue per time of day')
plt.xlabel('Day')
plt.ylabel('Revenue (Units)')

# 4. Display the chart
plt.show()
```
